```python
import matplotlib.pyplot as plt

def compound_interest(principal, rate, time):
    amount = principal * (1 + rate) ** time
    interest = amount - principal
    return interest

li = list(range(1,50)) # Liste des années de 1 à 49
initial_investment = 1000
interest_rate = 0.08

# Stocker les résultats d'intérêts dans une liste
interests = [compound_interest(initial_investment, interest_rate, years) for years in li]

# Tracer la courbe
plt.figure(figsize=(8, 5))
plt.plot(li, interests, marker='o', linestyle='-', color="#b41fa8", linewidth=2, label='Compound Interest')

# Personnalisation
plt.title("Evolution of Compound Interest ($1 000 at 8%)", fontsize=13, fontweight='bold')
plt.xlabel("Years", fontsize=11)
plt.ylabel("Interest Generated ($)", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

```python
import plotly.graph_objects as go

def compound_interest(principal, rate, time):
    amount = principal * (1 + rate) ** time
    interest = amount - principal
    return interest

li = list(range(1, 50))
initial_investment = 10000
interest_rate = 0.08

# Calcul des intérêts
interests = [compound_interest(initial_investment, interest_rate, years) for years in li]

# Création du tracé interactif
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=li,
    y=interests,
    mode='lines+markers',
    name='Compound Interest',
    hovertemplate='Year: %{x}<br>Interest: %{y:$,.0f}<extra></extra>',
    line=dict(color="#941fb4", width=2),
    marker=dict(size=7)
))

# Personnalisation de la disposition
fig.update_layout(
    title=f"Evolution of Compound Interest (${initial_investment:,} at {interest_rate*100:.0f}%)",
    xaxis_title="Years",
    yaxis_title="Interest Generated ($)",
    template="plotly_white",
    hovermode="x unified"
)

fig.show()
```
