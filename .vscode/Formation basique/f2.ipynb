```python
i_capital = 1000
f_capital = 2000
interest_rate = 0.02
years = 0

while i_capital < f_capital:
    i_capital += i_capital * interest_rate
    years += 1

print(f"It will take {years} years for the initial capital to double")
```

```python
for i in range(1 , 9) :
    i_capital = 1000
    f_capital = 2000
    interest_rate = i / 100
    years = 0

    while i_capital < f_capital:
        i_capital += i_capital * interest_rate
        years += 1

    print(f"At an interest rate of {interest_rate*100}%, it will take {years} years for the initial capital to double")
```
