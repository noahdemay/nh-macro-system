```python
a = 60
a1 = 20
b = 145
b1 = 10
c = 120
c1 = 8
d = 135
d1 = 22
```

```python
apple_stock_price_1 = a
apple_stock_number_1 = a1
apple_stock_value_1 = apple_stock_price_1 * apple_stock_number_1
print(apple_stock_value_1)

apple_stock_price_2 = b
apple_stock_number_2 = b1
apple_stock_value_2 = apple_stock_price_2 * apple_stock_number_2
print(apple_stock_value_2)

apple_stock_price_3 = c
apple_stock_number_3 = c1

apple_stock_price_4 = d
apple_stock_number_4 = d1

apple_stock_value_4 = apple_stock_price_4 * apple_stock_number_4
print(apple_stock_value_4)

#John a actuellement un portefeuille d'actions valorisé à 2970$ .
```

```python
apple_stock_price_average = (apple_stock_price_1 * apple_stock_number_1 + apple_stock_price_2 * apple_stock_number_2) / (apple_stock_number_1 + apple_stock_number_2)
print(apple_stock_price_average)

#John a acheté ses actions à un prix moyen de 88,3$.
```

```python
apple_stock_operation_1 = apple_stock_price_1 * -1 * apple_stock_number_1
apple_stock_operation_2 = apple_stock_price_2 * -1 * apple_stock_number_2
apple_stock_operation_3 = apple_stock_price_3 * apple_stock_number_3
apple_stock_operation_4 = apple_stock_price_4 * apple_stock_number_4

apple_stock_balance = apple_stock_operation_1 + apple_stock_operation_2 + apple_stock_operation_3 + apple_stock_operation_4
print(apple_stock_balance)

#si John vend toutes ses actions restantes, il aura un bénéfice de 1280$ depuis le début de son investissement.
```

```python
tax_france = apple_stock_balance * 0.30
tax_france > 0
print(tax_france)
tax_belgium = (0 * (500/apple_stock_balance))+ (0.4 * (apple_stock_balance - 500))
tax_belgium > 0
print(tax_belgium)

#John paiera 384$ d'impots en France et 312$ d'impôts en Belgique. 
#Il est donc plus avantageux pour John de vendre ses actions en Belgique.
```
